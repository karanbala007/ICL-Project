# Word Embeddings Demo

This notebook demonstrates two key concepts in NLP:
- One-hot encoding for simple word representation
- Word2Vec embeddings for learning semantic similarity from context

Suggested improvements for a stronger project:
1. Use a larger text corpus instead of a tiny toy dataset.
2. Compare Word2Vec with GloVe or FastText.
3. Tune vector_size, window, and epochs to improve word similarity quality.
4. Visualize embeddings with t-SNE or PCA for better interpretation.


```python
!pip install -q scikit-learn gensim scipy

from sklearn.preprocessing import OneHotEncoder
from gensim.models import Word2Vec
import numpy as np

words = [['cat'], ['dog'], ['apple'], ['car']]

encoder = OneHotEncoder(sparse_output=False)
result = encoder.fit_transform(words)

print("Words:", encoder.categories_[0])
print("\nOne-Hot Embeddings:")
print(result)
print("\nEmbedding matrix shape:", result.shape)
```

```python
sentences = [
    ["python", "is", "a", "language"],
    ["java", "is", "a", "language"],
    ["python", "is", "easy"],
    ["java", "is", "easy"],
    ["python", "is", "powerful"],
    ["java", "is", "powerful"]
]

model = Word2Vec(
    sentences=sentences,
    vector_size=10,
    window=2,
    min_count=1,
    workers=1,
    seed=42,
    epochs=100
)

print("\nWords similar to 'java':")
print(model.wv.most_similar("java", topn=3))

print("\nWords similar to 'python':")
print(model.wv.most_similar("python", topn=4))

print("\nWord vector for 'python':")
print(model.wv['python'])
print("\nVocabulary size:", len(model.wv.key_to_index))
```

## Key Takeaways

- One-hot encoding creates binary vectors that represent words without any semantic relationship.
- Word2Vec embedding captures context-driven similarity between words.
- Embeddings are useful for downstream tasks like text classification, clustering, and recommendation.
