# RAG Step 1: Document Loading and Chunking

This notebook demonstrates the first step in a RAG pipeline: loading source text and splitting it into meaningful chunks for embeddings and retrieval.

We keep the workflow simple and easy to understand for beginners.


```python
!pip install -q chromadb sentence-transformers

from textwrap import TextWrapper

source_documents = [
    "RAG stands for Retrieval-Augmented Generation. It combines a retriever with a language model to answer questions using external knowledge.",
    "In a typical RAG workflow, a document is chunked, embedded, stored in a vector database, and then retrieved based on semantic similarity.",
    "ChromaDB is a popular vector database that stores embeddings and supports similarity search using metadata and filters.",
    "Large language models can hallucinate when they do not have enough context. RAG reduces this by grounding responses in relevant retrieved passages.",
    "The process usually includes loading documents, splitting them into chunks, embedding the chunks, storing them, and then querying the database."
]

def chunk_text(text, chunk_size=250, overlap=40):
    wrapper = TextWrapper(width=chunk_size, break_long_words=False, break_on_hyphens=False, replace_whitespace=False)
    lines = text.splitlines()
    paragraphs = []
    current = ''
    for line in lines:
        if line.strip():
            current += line.strip() + ' '
        else:
            if current.strip():
                paragraphs.append(current.strip())
                current = ''
    if current.strip():
        paragraphs.append(current.strip())

    chunks = []
    for paragraph in paragraphs:
        words = paragraph.split()
        for i in range(0, len(words), chunk_size - overlap):
            chunk = ' '.join(words[i:i + chunk_size])
            if chunk:
                chunks.append(chunk)
    return chunks

all_chunks = []
for i, doc in enumerate(source_documents, start=1):
    chunks = chunk_text(doc)
    for j, chunk in enumerate(chunks, start=1):
        all_chunks.append({
            'doc_id': f'doc_{i}',
            'chunk_id': f'doc_{i}_chunk_{j}',
            'text': chunk
        })

print(f'Total documents: {len(source_documents)}')
print(f'Total chunks created: {len(all_chunks)}')
print('--- First few chunks ---')
for item in all_chunks[:3]:
    print(item['chunk_id'], '->', item['text'])
```

## Why this matters

Chunking matters because large documents are often too long for embedding models and LLMs to handle efficiently. Smaller chunks keep the retrieval process focused and improve semantic search quality.
